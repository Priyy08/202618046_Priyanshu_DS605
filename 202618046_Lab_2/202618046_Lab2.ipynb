Task 2 : Vectorized Arithmetic and Linear Algebra

```python
## 1. Create two matrices and perform addition, element-wise multiplication, and matrix multiplication

import numpy as np

# Create two 2x2 matrices
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

# Matrix Addition
addition = A + B
print("\nMatrix Addition (A + B):")
print(addition)

# Element-wise Multiplication
elementwise_multiplication = A * B
print("\nElement-wise Multiplication (A * B):")
print(elementwise_multiplication)

# Matrix Multiplication
matrix_multiplication = A @ B
print("\nMatrix Multiplication (A @ B):")
print(matrix_multiplication)
```

```python
## 2. Find the transpose and determinant of a square matrix

import numpy as np

# Create a square matrix
A = np.array([
    [1, 2],
    [3, 4]
])

print("Matrix A:")
print(A)

# Transpose
transpose = A.T

print("\nTranspose of A:")
print(transpose)

# Determinant
determinant = np.linalg.det(A)

print("\nDeterminant of A:")
print(determinant)
```

```python
## 3. Find the inverse if the matrix is invertible and verify A @ A⁻¹

import numpy as np

# Create a square matrix
A = np.array([
    [1, 2],
    [3, 4]
])

print("Matrix A:")
print(A)

# Calculate determinant
determinant = np.linalg.det(A)

print("\nDeterminant:", determinant)

# Check whether matrix is invertible
if not np.isclose(determinant, 0):
    # Find inverse
    A_inverse = np.linalg.inv(A)

    print("\nInverse of A:")
    print(A_inverse)

    # Verify A @ A^-1
    verification = A @ A_inverse

    print("\nA @ A^-1:")
    print(verification)

    # Check whether result is approximately identity matrix
    identity_matrix = np.eye(A.shape[0])

    print("\nIs A @ A^-1 equal to the identity matrix?")
    print(np.allclose(verification, identity_matrix))

else:
    print("\nMatrix A is not invertible.")
```

```python
## 4. Vectorized NumPy operations — no explicit Python loops

import numpy as np

# Create two matrices
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [6, 5, 4],
    [3, 2, 1]
])

# All operations are vectorized
addition = A + B
subtraction = A - B
multiplication = A * B
squared_A = A ** 2
square_root_A = np.sqrt(A)

print("A + B:")
print(addition)

print("\nA - B:")
print(subtraction)

print("\nElement-wise A * B:")
print(multiplication)

print("\nA squared:")
print(squared_A)

print("\nSquare root of A:")
print(square_root_A)
```

```python
## 1. Generate at least 1,000 values from a normal distribution

import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Chosen parameters
mean = 50
std_dev = 10
num_values = 1000

# Generate 1000 values from a normal distribution
data = np.random.normal(
    loc=mean,
    scale=std_dev,
    size=num_values
)

print("Chosen Mean:", mean)
print("Chosen Standard Deviation:", std_dev)
print("Number of Values:", len(data))

print("\nFirst 10 generated values:")
print(data[:10])
```

**Task 5 - Filtering and Querying**


```python
## 1. Male passengers older than 50: how many?

import pandas as pd

# Load Titanic dataset
df = pd.read_csv("train.csv")

# Filter male passengers older than 50
male_over_50 = df[(df["Sex"] == "male") & (df["Age"] > 50)]

# Count passengers
count_male_over_50 = len(male_over_50)

print("Male passengers older than 50:", count_male_over_50)
```

```python
##2.Female first-class passengers: how many, and what percentage survived?

# Filter female passengers in first class
female_first_class = df[
    (df["Sex"] == "female") &
    (df["Pclass"] == 1)
]

# Total number of female first-class passengers
total_female_first_class = len(female_first_class)

# Number who survived
survived_female_first_class = female_first_class["Survived"].sum()

# Survival percentage
survival_percentage = (
    survived_female_first_class / total_female_first_class
) * 100

print("Female first-class passengers:", total_female_first_class)
print("Female first-class passengers who survived:",
      survived_female_first_class)
print("Percentage survived:", survival_percentage, "%")
```

```python
## 3. 3. Age 20–40, Fare above overall median, and survived — how many?

# Calculate overall median fare
overall_median_fare = df["Fare"].median()

print("Overall median Fare:", overall_median_fare)

# Filter passengers
filtered_passengers = df[
    (df["Age"].between(20, 40)) &
    (df["Fare"] > overall_median_fare) &
    (df["Survived"] == 1)
]

# Count passengers
count_filtered = len(filtered_passengers)

print("Passengers meeting all conditions:", count_filtered)
```

```python
## 4. Travelling alone, age < 30, and did not survive — how many?

# Filter passengers travelling alone, younger than 30,
# and who did not survive
alone_under_30_not_survived = df[
    (df["SibSp"] == 0) &
    (df["Parch"] == 0) &
    (df["Age"] < 30) &
    (df["Survived"] == 0)
]

# Count passengers
count_alone = len(alone_under_30_not_survived)

print("Passengers travelling alone, age < 30, and did not survive:",
      count_alone)
```

```python
## 5. Embarked = 'S', Pclass 2 or 3, and Fare above Southampton median — how many?

# Calculate median Fare for Southampton passengers
southampton_median_fare = df.loc[
    df["Embarked"] == "S",
    "Fare"
].median()

print("Southampton median Fare:", southampton_median_fare)

# Filter:
# Embarked = S
# Pclass = 2 or 3
# Fare > Southampton median
filtered_southampton = df[
    (df["Embarked"] == "S") &
    (df["Pclass"].isin([2, 3])) &
    (df["Fare"] > southampton_median_fare)
]

# Count passengers
count_southampton = len(filtered_southampton)

print("Passengers meeting all conditions:", count_southampton)
```

**Task 6 - groupby() and Aggregation**

```python
## 1.Survival rate by Sex.

import pandas as pd

# Load Titanic dataset
df = pd.read_csv("train.csv")

# Survival rate by Sex
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100

print("Survival Rate by Sex:")
print(survival_by_sex)
```

```python
## 2. Survival rate by Pclass

# Survival rate by Passenger Class
survival_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100

print("Survival Rate by Pclass:")
print(survival_by_pclass)
```

```python
## 3. Average Age and Fare by Pclass

# Average Age and Fare for each Passenger Class
average_by_pclass = df.groupby("Pclass")[["Age", "Fare"]].mean()

print("Average Age and Fare by Pclass:")
print(average_by_pclass)
```

```python
## 4. Passenger count and survival rate by Sex-Pclass

# Group by Sex and Pclass
sex_pclass_summary = df.groupby(
    ["Sex", "Pclass"]
).agg(
    Passenger_Count=("PassengerId", "count"),
    Survival_Rate=("Survived", "mean")
)

# Convert survival rate to percentage
sex_pclass_summary["Survival_Rate"] = (
    sex_pclass_summary["Survival_Rate"] * 100
)

print("Passenger Count and Survival Rate by Sex-Pclass:")
print(sex_pclass_summary)
```

```python
## 5. Passenger count, average Fare, and survival rate by Embarked

# Group by Embarked
embarked_summary = df.groupby("Embarked").agg(
    Passenger_Count=("PassengerId", "count"),
    Average_Fare=("Fare", "mean"),
    Survival_Rate=("Survived", "mean")
)

# Convert survival rate to percentage
embarked_summary["Survival_Rate"] = (
    embarked_summary["Survival_Rate"] * 100
)

print("Passenger Count, Average Fare, and Survival Rate by Embarked:")
print(embarked_summary)
```

**Task 7 - Missing Values and Fare Outliers**

```python
## 1. Missing-value count and percentage for every column + bar chart

import pandas as pd
import matplotlib.pyplot as plt

# Load Titanic dataset
df = pd.read_csv("train.csv")

# Calculate missing-value count
missing_count = df.isnull().sum()

# Calculate missing-value percentage
missing_percentage = (missing_count / len(df)) * 100

# Create a summary table
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("Missing Value Summary:")
print(missing_summary.round(2))
```

```python
## Plot missing-value counts

# Plot missing-value counts
missing_count.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Missing Values by Column")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
## 2. Fill missing Age values with the mean Age

# Count missing Age values before filling
age_missing_before = df["Age"].isnull().sum()

print("Missing Age values before filling:",
      age_missing_before)

# Calculate mean Age
mean_age = df["Age"].mean()

print("Mean Age:", mean_age)

# Fill missing Age values with mean Age
df["Age"] = df["Age"].fillna(mean_age)

# Count missing Age values after filling
age_missing_after = df["Age"].isnull().sum()

print("Missing Age values after filling:",
      age_missing_after)
```

```python
## 3. Impute NaN values using mean, median, mode, and random value

## A. Mean Imputation
# Reload original dataset so Age contains NaN values again
df = pd.read_csv("train.csv")

# Mean imputation
mean_age = df["Age"].mean()

df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(mean_age)

print("Mean imputation:")
print("Missing Age values:",
      df_mean["Age"].isnull().sum())
```

```python
## B. Median Imputation

median_age = df["Age"].median()

df_median = df.copy()
df_median["Age"] = df_median["Age"].fillna(median_age)

print("Median imputation:")
print("Median Age:", median_age)
print("Missing Age values:",
      df_median["Age"].isnull().sum())
```

```python
## C. Mode Imputation
mode_age = df["Age"].mode()[0]

df_mode = df.copy()
df_mode["Age"] = df_mode["Age"].fillna(mode_age)

print("Mode imputation:")
print("Mode Age:", mode_age)
print("Missing Age values:",
      df_mode["Age"].isnull().sum())
```

```python
## D. Random-value imputation
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

df_random = df.copy()

# Find positions where Age is missing
missing_positions = df_random["Age"].isnull()

# Get existing non-missing Age values
available_ages = df_random.loc[
    df_random["Age"].notnull(),
    "Age"
].values

# Generate random Age values
random_ages = np.random.choice(
    available_ages,
    size=missing_positions.sum(),
    replace=True
)

# Fill missing Age values
df_random.loc[missing_positions, "Age"] = random_ages

print("Random-value imputation:")
print("Missing Age values:",
      df_random["Age"].isnull().sum()) 
```

```python
## 4. Calculate Q1, Q3, IQR, 1.5 × IQR bounds, and Fare outliers

# Work with Fare column
fare = df["Fare"]

# Calculate Q1 and Q3
Q1 = fare.quantile(0.25)
Q3 = fare.quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate lower and upper bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Find outliers
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]

# Count outliers
outlier_count = len(fare_outliers)

print("Fare Outlier Analysis")
print("---------------------")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("1.5 × IQR:", 1.5 * IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Fare Outliers:", outlier_count)
```

```python
## 1. Create FamilySize and IsAlone

import pandas as pd

# Load Titanic dataset
df = pd.read_csv("train.csv")

# Create FamilySize
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1

# Create IsAlone
# IsAlone = 1 when FamilySize is 1, otherwise 0
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

# Display the new columns
print("FamilySize and IsAlone:")
print(df[["PassengerId", "SibSp", "Parch", "FamilySize", "IsAlone"]].head(10))
```

```python
## 3. Display the pivot table as percentages

# Convert survival rates to percentages
pivot_percentage = pivot_table * 100

print("Survival Rate Pivot Table (%):")
print(pivot_percentage.round(2))
```

**Task 9 - Visualizations and Observations**

```python
## 1. Correlation heatmap for relevant numerical columns

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Titanic dataset
df = pd.read_csv("train.csv")

# Select relevant numerical columns
numerical_columns = [
    "Survived",
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare"
]

correlation = df[numerical_columns].corr()

# Plot correlation heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Titanic Dataset - Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
## Find the strongest positive and negative relationships

# Remove self-correlations from the correlation matrix
corr_matrix = correlation.copy()

# Replace diagonal values with NaN
import numpy as np

np.fill_diagonal(corr_matrix.values, np.nan)

# Find strongest positive correlation
positive_pair = corr_matrix.stack().idxmax()
positive_value = corr_matrix.stack().max()

# Find strongest negative correlation
negative_pair = corr_matrix.stack().idxmin()
negative_value = corr_matrix.stack().min()

print("Strongest Positive Relationship:")
print(
    positive_pair[0],
    "and",
    positive_pair[1],
    "->",
    round(positive_value, 3)
)

print("\nStrongest Negative Relationship:")
print(
    negative_pair[0],
    "and",
    negative_pair[1],
    "->",
    round(negative_value, 3)
)
```

```python
## 2. Plot survival rate by Sex

# Calculate survival rate by Sex
survival_by_sex = (
    df.groupby("Sex")["Survived"].mean() * 100
)

print("Survival Rate by Sex:")
print(survival_by_sex.round(2))
```
